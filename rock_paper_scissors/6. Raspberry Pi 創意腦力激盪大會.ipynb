# 🧠 6. Raspberry Pi 創意腦力激盪大會 💡

Raspberry Pi (樹莓派) 就像一個大腦，它可以連接各種感測器 (Sensors) 當作眼睛和耳朵，也可以控制馬達、燈光或發送訊息。 現在，請你們分組，利用接下來的時間，構思一個你們最想實現的 Project。

---

## 🚀 第一階段：靈感觸發 (Themes)

如果你們一時沒有頭緒，可以從以下幾個方向開始思考：

* 🏠 **智慧家居 (Smart Home):** 怎樣讓宿舍或房間更聰明？(例如：自動澆花機、智能鏡子、自動開關燈)
* 🎮 **娛樂與遊戲 (Gaming):** 自製復古遊戲機、光劍控制器、互動式音樂盒。
* 🌍 **社會與環境 (Social Good):** 噪音監測器、空氣品質警報、幫助視障人士的輔助工具。
* 🤖 **AI 與監控 (AI & Monitor):** 寵物監視器 (看到狗狗就拍照)、人臉辨識門鎖。

---

## ❓ 第二階段：關鍵思考題 (Guiding Questions)

請在小組討論中，嘗試回答以下 5 個問題。這將幫助你們把模糊的想法變成具體的計劃：

### 1. 你想解決什麼問題？(The "Why")
不需要很偉大，可以很微小。
* 是因為懶得起床關燈？
* 是因為總是忘記帶鑰匙？
* 還是純粹覺得某樣東西很酷，想做出來炫耀？

### 2. 你的 Pi 需要「感覺」什麼？(Input)
Raspberry Pi 需要什麼資訊才能運作？
* 它需要看到東西嗎？(相機 Camera Module)
* 它需要知道溫度、濕度或是距離嗎？(Sensors)
* 它需要聽到聲音或是透過按鈕觸發嗎？

### 3. 你的 Pi 會做出什麼「反應」？(Output)
接收到訊號後，它要做什麼？
* **視覺：** 在螢幕上顯示文字？閃爍 LED 燈？
* **動作：** 啟動馬達轉動？打開繼電器 (Relay) 控制家電？
* **通訊：** 發送 Line/Telegram 通知給你？發出一封 Email？

### 4. 這個裝置是給誰用的？(User Experience)
* 如果是給阿嬤用的，按鈕是不是要大一點？
* 如果是給寵物用的，外殼是不是要防水、防咬？

### 5. 這個 Project 最「酷」的地方在哪裡？(The Unique Selling Point)
為什麼我們要選你們這組的點子？有一句話可以形容它的賣點嗎？

---

## 🎤 第三階段：分享時間 (Pitch Time)

請每組推派一位代表，用 **2 分鐘** 的時間向全班分享。請依照以下格式報告：

> **1. 專案名稱：** (取一個響亮的名字)
>
> **2. 我們發現的問題是：** ...
>
> **3. 我們的解決方案是：** 利用 Raspberry Pi 偵測 [輸入訊號]，然後 [輸出動作]。
>
> **4. 我們覺得最有趣的地方是：** ...

---

### 👩‍🏫 老師的小提示 (Teacher's Note)

* **不要擔心技術難度：** 先天馬行空地想，我們之後再來討論可行性 (Feasibility)。
* **善用 "If This Then That" 的邏輯：** 例如「如果 **(If)** 偵測到人走過，那麼 **(Then)** 就拍照傳到我的手機」。
